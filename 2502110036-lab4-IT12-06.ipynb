IT 10

```python
import random

accept_count = 0
rework_count = 0
scrap_count = 0

for i in range(5):
    diameter = random.uniform(2, 25)

    if 20 <= diameter <= 25:
        accept_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> ACCEPT")
    elif 20 <= diameter < 20 or 25 < diameter <= 25:
        rework_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> REWORK")
    else:
        scrap_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> SCRAP")


print("\nSHIFT REPORT")
print(f"ACCEPT: {accept_count}")
print(f"REWORK: {rework_count}")
print(f"SCRAP:  {scrap_count}")
```

IT 8

```python
import random

mode = input("Enter mode (halt/skip): ").strip().lower()

accept_count = 0
rework_count = 0
scrap_count = 0
inspected = 0

for i in range(4):
    diameter = random.uniform(15, 30)
    inspected += 1

    if 15 <= diameter <= 25:
        accept_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> ACCEPT")

    elif 15 <= diameter < 20 or 20 < diameter <= 22:
        rework_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> REWORK")

    else:
        scrap_count += 1
        print(f"Part {i + 1}: {diameter:.2f} mm -> SCRAP")

        if mode == "halt":
            print(f"Batch halted after {inspected} parts.")
            break

        elif mode == "skip":
            print("SCRAP logged; continuing batch.")
            continue

        else:
            print("Unknown mode. Batch stopped.")
            break

print("\nSHIFT REPORT")
print(f"Inspected: {inspected}")
print(f"ACCEPT: {accept_count}")
print(f"REWORK: {rework_count}")
print(f"SCRAP: {scrap_count}")
```

IT 6

```python
import random

consecutive_accepts = 0
attempts = 0
target = 5
safety_cap = 500

while consecutive_accepts < target and attempts < safety_cap:
    attempts += 1
    diameter = random.uniform(2, 25)

    if 2 <= diameter <= 20:
        consecutive_accepts += 1
        print(f"Part {attempts}: {diameter:.2f} mm -> ACCEPT "
            f"| consecutive ACCEPTs: {consecutive_accepts}")

    elif 2 <= diameter < 2.5 or 2.2 < diameter <= 20:
        consecutive_accepts = 0
        print(f"Part {attempts}: {diameter:.2f} mm -> REWORK")

    else:
        consecutive_accepts = 0
        print(f"Part {attempts}: {diameter:.2f} mm -> SCRAP")

if consecutive_accepts == target:
    print(f"\nFive consecutive ACCEPTs achieved in {attempts} parts.")
else:
    print(f"\nSafety cap reached after {attempts} attempts.")
    print("Five consecutive ACCEPTs were not achieved.")
```
