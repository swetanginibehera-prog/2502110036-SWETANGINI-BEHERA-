```python
def tally(labels):
    counts = {}

    for label in labels:
        counts[label] = counts.get(label, 0) + 1

    return counts


labels = ["ACCEPT", "REWORK", "ACCEPT", "SCRAP"]

print(tally(labels))
```

```python
def lookup(catalogue, code):
    return catalogue.get(code, "Part not found")


def scrap_codes(catalogue):
    return {
        code for code, dia in catalogue.items()
        if inspect(dia) == "SCRAP"
    }


catalogue = {
    "P01": 25.00,
    "P02": 24.92,
    "P03": 25.20,
    "P04": 25.03
}

print("P01:", lookup(catalogue, "P01"))
print("P10:", lookup(catalogue, "P10"))
print("Scrap codes:", scrap_codes(catalogue))
```

cell 1- create shoptools.py

```python
%%writefile shoptools.py

def inspect(dia):
    if 24.95 <= dia <= 25.05:
        return "ACCEPT"
    elif 24.90 <= dia <= 25.10:
        return "REWORK"
    else:
        return "SCRAP"


def tally(labels):
    counts = {}

    for label in labels:
        counts[label] = counts.get(label, 0) + 1

    return counts


def report(shift, counts):
    return f"Shift {shift}: {counts}"


def lookup(catalogue, code):
    return catalogue.get(code, "Part not found")


def scrap_codes(catalogue):
    return {
        code for code, dia in catalogue.items()
        if inspect(dia) == "SCRAP"
    }
```

cell-2

```python
import shoptools

logs = ["P01,25.00", "P02,24.92", "P03,25.20", "P04,25.03"]

catalogue = {}
labels = []

for line in logs:
    code, dia = line.split(",")
    dia = float(dia)
    catalogue[code] = dia
    labels.append(shoptools.inspect(dia))

counts = shoptools.tally(labels)
print(shoptools.report("A", counts))
print("P01:", shoptools.lookup(catalogue, "P01"))
print("Scrap:", shoptools.scrap_codes(catalogue))
```

```python
import random

def inspect(dia):
    if 24.95 <= dia <= 25.05:
        return "ACCEPT"
    elif dia > 25.05:
        return "REWORK"
    else:
        return "SCRAP"

def tally(labels):
    counts = {"ACCEPT": 0, "REWORK": 0, "SCRAP": 0}
    for label in labels:
        counts[label] += 1
    return counts

def report(shift, counts):
    print("Shift:", shift)
    print("ACCEPT:", counts["ACCEPT"])
    print("REWORK:", counts["REWORK"])
    print("SCRAP:", counts["SCRAP"])

labels = []

for i in range(10):
    dia = round(random.uniform(24.90, 25.10), 2)
    label = inspect(dia)
    print(dia, "->", label)
    labels.append(label)

counts = tally(labels)
report("A", counts)
```
