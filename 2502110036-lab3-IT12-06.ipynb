IT 10

```python
accept = 0
rework = 0
scrap = 0

for i in range(5):
  diameter = float(input("Enter diameter: "))
  if 10 <= diameter <= 25:
    print("ACCEPT")
    accept += 1
  elif(10 <= diameter < 25)or(25 < diameter <= 25.10):
    print("REWORK")
    rework += 1
  else:
    print("SCRAP")
    scrap += 1

print()
print("Accepted =", accept)
print("Rework =", rework)
print("Scrap =", scrap)
```

```python
total = int(input("Enter total parts: "))
size = int(input("Enter box size: "))

boxes = total // size
loose = total % size

print("Full boxes:", boxes)
print("Loose parts:", loose)

print("Audit:", boxes * size + loose == total)
```

IT 6

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    dia = float(input("Enter diameter: "))

    if 9.8 <= dia <= 10.2:
        accept += 1
    elif 9.5 <= dia < 9.8 or 10.2 < dia <= 10.5:
        rework += 1
    else:
        scrap += 1

boxes = accept // 4
loose = accept % 4

print("End of Shift Report")
print("Accepted:", accept)
print("Rework:", rework)
print("Scrap:", scrap)
print("Full boxes:", boxes)
print("Loose accepted parts:", loose)
```

```python
def add_numbers(*args):
  """Adds an arbitary number of arguments."""
  total = 0
  for num in args:
    total += num
  return total

#Example usage:
result = add_numbers(1,2,3,4,5)
print(f"Sum of numbers (1,2,3,4,5):{result}")
```
