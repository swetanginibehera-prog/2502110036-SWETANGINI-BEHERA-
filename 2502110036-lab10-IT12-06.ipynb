```python
from pathlib import Path
from datetime import datetime

try:
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    BASE_DIR = Path.cwd()

DATA_FILE = BASE_DIR / "data" / "sample.txt"


def load():
    records = []

    with open(DATA_FILE, "r") as file:
        for line_no, line in enumerate(file, start=1):
            parts = line.strip().split("|")

            if len(parts) != 5:
                raise ValueError(f"Invalid record at line {line_no}")

            equipment_id, name, date, issue, hours = parts

            datetime.strptime(date, "%Y-%m-%d")

            hours = float(hours)

            if hours < 0 or hours > 100:
                raise ValueError(f"Invalid downtime at line {line_no}")

            records.append({
                "id": equipment_id,
                "name": name,
                "date": date,
                "issue": issue,
                "hours": hours
            })

    return records


def summarise(records):
    total = sum(record["hours"] for record in records)

    print("\nMaintenance Report")
    print("------------------")
    print("Total Records:", len(records))
    print("Total Downtime:", total, "hours")
```
